## Telco Customer Churn Prediction

**Project Type:** Machine Learning / Classification

**Objective:** Predict which telecommunications customers are likely to churn and identify the factors associated with customer attrition.

---

### Business Problem

Customer churn is a major business problem for telecommunications companies because acquiring a new customer can be more expensive than retaining an existing one.

The goal of this project is to develop a machine learning system that predicts whether a customer is likely to churn based on their demographic information, account characteristics, service usage, contract information, and billing details.

The model should provide reliable churn predictions that can support proactive customer-retention strategies.

---

### Expected Outcome

The project will produce:

1. A validated and reproducible data-processing pipeline.
2. A machine learning classification model for churn prediction.
3. Model evaluation using business-relevant classification metrics.
4. Feature-level insights explaining factors associated with churn.
5. A reusable prediction pipeline suitable for deployment.

## 1. Business Problem

### Problem Statement

The telecommunications company wants to reduce customer churn by identifying customers who are at high risk of leaving the company.

Currently, customer-retention efforts may be inefficient if they are applied broadly to all customers. A predictive model can help prioritize customers who are more likely to churn.

### Business Question

> Which customers are most likely to churn, and can machine learning identify them early enough to support targeted retention actions?

### Business Impact

A successful churn prediction system could help the business:

- Identify high-risk customers.
- Prioritize retention campaigns.
- Improve customer retention.
- Reduce unnecessary retention costs.
- Better understand factors associated with customer churn.
- Support data-driven customer relationship management.

## 2. Business Assumptions

The project assumes that:

- Customer information is available before the churn decision occurs.
- Historical customer records are representative of future customers.
- The features available in the dataset can be obtained during normal business operations.
- Churn predictions can be generated early enough for the business to intervene.
- Customers identified as high-risk can be targeted with appropriate retention strategies.
- The cost of false negatives is potentially higher than the cost of some false positives because failing to identify a customer who churns may result in lost revenue.

These assumptions should be validated before deploying the model in a real business environment.

## 7. Machine Learning Problem Formulation

### Problem Type

This is a **binary classification** problem.

### Input

Customer-level demographic, service, account, contract, and billing information.

### Output

A prediction indicating whether the customer is likely to churn.

### Target Encoding

| Original | Encoded |
|---|---:|
| No | 0 |
| Yes | 1 |

### Prediction Outputs

The final system should ideally provide:

- Predicted class (`0` or `1`)
- Probability of churn
- A configurable decision threshold

The probability output is important because the business may want to prioritize customers by risk rather than simply classify them as churn/non-churn.

## 8. Success Metrics

Because customer churn is a binary classification problem and the target may be imbalanced, model performance will not be evaluated using accuracy alone.

### Primary Metrics

#### Recall

Measures the proportion of actual churners correctly identified by the model.

High recall is important because failing to identify a customer who eventually churns represents a potential lost customer.

#### ROC-AUC

Measures the model's ability to distinguish between customers who churn and customers who do not churn across classification thresholds.

#### PR-AUC

Precision-Recall AUC will also be considered because it can provide useful information when the positive churn class is relatively less frequent.

### Secondary Metrics

- Precision
- F1-score
- Accuracy
- Confusion matrix
- Calibration / probability quality where appropriate

### Business-Oriented Evaluation

The final model should also be evaluated using:

- Churn capture rate among the highest-risk customers.
- Precision among customers targeted for intervention.
- Prediction threshold analysis.
- Potential retention cost versus potential churn loss.

The final decision threshold should therefore be based on business constraints rather than automatically using 0.50.

## 9. Baseline Strategy

A baseline model will be established before evaluating more sophisticated algorithms.

Possible baselines include:

1. Majority-class classifier.
2. Logistic Regression.
3. Tree-based classification model.

The baseline provides a reference point against which more complex models can be evaluated.

A more complex model will only be preferred if it provides meaningful improvement in predictive performance, robustness, interpretability, or business value.

Models will be evaluated in order of increasing complexity, ensuring that added sophistication is justified by measurable performance gains.

Each successive model must demonstrate a meaningful improvement in the primary metrics (Recall, ROC-AUC) to be carried forward into final model selection.

## 11. Data Leakage Considerations

The following practices will be used to minimize data leakage:

- The test set will remain untouched until final evaluation.
- Preprocessing transformations will be fitted only on the training data.
- Validation and test data will only be transformed using fitted training preprocessing.
- Feature engineering will avoid using information that would not be available at prediction time.
- Target information will not be used as an input feature.
- Customer identifiers will not be used as predictive features.
- Resampling techniques, if required, will be applied only to the training data.
